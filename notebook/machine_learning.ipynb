```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split, RandomizedSearchCV
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

from xgboost import XGBRegressor

import pickle
```

```python
#Let's do some feature engineering

df = pd.read_csv("demand_forecasting.csv")
```

```python
df
```

```python
df.columns
```

```python
features = [
    "Price",
    "Discount",
    "Inventory Level",
    "Promotion",
    "Competitor Pricing",
    "Category"
]
target = "Demand"
```

```python
X = df[features].copy()
```

```python
X
```

```python
y = df[target]
```

```python
y
```

```python
label_encoders = {}

categorical_cols = X.select_dtypes(include = "object").columns
```

```python
categorical_cols
```

```python
for col in categorical_cols:
    le = LabelEncoder()
    X[col] = le.fit_transform(X[col])
    label_encoders[col] = le
```

```python
label_encoders
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2)
```

```python
#We'll skip scaling because it is not requried in tree based algorithms

xgb = XGBRegressor(objective = "reg:squarederror", n_jobs = -1)
```

```python
#To get the best performance from our model, we'll do hyperparameter tuning

param_dict = {
    "n_estimators" : [200, 300, 500],
    "max_depth" : [3, 4, 6, 8],
    "learning_rate" : [0.01, 0.05, 0.1],
    "subsample" : [0.7, 0.8, 1.0],
    "colsample_bytree" : [0.7, 0.8, 1.0],
    "min_child_weight" : [1,3,5]
}
```

```python
random_search = RandomizedSearchCV(
    estimator = xgb,
    param_distributions = param_dict,
    n_iter = 25,
    scoring = "neg_mean_absolute_error",
    cv = 3,
    verbose = 1,
    n_jobs = -1
)
```

```python
random_search.fit(X_train, y_train)
```

```python
random_search.best_params_
```

```python
best_model = random_search.best_estimator_
y_pred = best_model.predict(X_test)
```

```python
mse = mean_squared_error(y_test, y_pred)
print(np.sqrt(mse))  #root mean squared error
```

```python
best_model.feature_importances_
```

```python
feature_importance = pd.Series(
    best_model.feature_importances_,
    index = X.columns
).sort_values(ascending = False)
```

```python
feature_importance

#These are effectively the weights of each feature in the model. The higher the value, the more important the feature is in predicting the target variable.
```

```python
feature_importance.plot(kind = "bar", title = "Feature Importances")
```

```python
with open("label_encoders.pkl", "wb") as f:
    pickle.dump(label_encoders, f)
```
